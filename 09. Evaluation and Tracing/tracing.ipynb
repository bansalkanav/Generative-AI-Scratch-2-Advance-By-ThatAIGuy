## **What is tracing and why tracing?**

Tracing helps **record inputs and outputs** of intermediate steps and capture hierarchical call stack

**Note**
- GenAI applications are complex as the internal logic is opaque
- Without tracing, one failed LM call can break the whole program

**Tracing improves:**
- Interpretability
- Debugging

## **What is MLFlow?**
Reference: https://learn.deeplearning.ai/courses/dspy-build-optimize-agentic-apps/information

MLFlow is an opensource MLOps framework. Both ML Server and Client are Opensource.
- Supports ML and GenAI project development
- Covers the full lifecycle: development, tracking and deployment
- Focus on reproducibility and tracability
- Helps manage every experiment, run and model version with ease
