### **Goal of evaluating the model**

Have a quantitative response to:
1. Is the LLM improving?
2. Where is it hallucinating?

Evaluation helps us understand:
1. Is the system working?
2. As I improve the system, where can I take it?

A good evaluation:
1. Is quantitative i.e. measurable scores
2. Points at what you can improve i.e. sets up the next frontier correctly
3. Scalable and automated i.e. use LLMs to scale the evals


### **Practical tips for your evaluation dataset**
- Define easiest examples that still fails
- Adversarial Playground - Hook LLM with the playground and try to break it to understand the boundaries
- Set the next accuracy target for your LLM

**Summary**
1. Start with a small focused evaluation dataset.
2. Set the target accuracy.
3. Try to improve from 20% to 90%+
4. Update your evaluation set to make it harder.
5. So that once the evaluation drop to 20, you move your way back to 90+.

### **Evaluations are `iterative`**

1. As you improve the system, you'll continue to expand your evaluation dataset with more breadth.
    - Play with adversarial playground
    - Cover harder examples

2. Improve your scoring mechanism to catch mistakes
    - It can be using LLMs to do semantic matching
    - Or deterministic systems to do exact matches

3. Iteratively add harder hallucination examples, as the system learn to get the easier ones.

4. Be rigorous about tracking the evaluation results over iterations, so you know which configuration produce which result.
