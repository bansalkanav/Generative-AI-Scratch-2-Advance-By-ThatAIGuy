Then in your Python code, use DefaultAzureCredential which will automatically use your Azure CLI credentials: 
```python

credential = DefaultAzureCredential(

)
```

## **Embedding Model**

````python
from langchain_openai import AzureOpenAIEmbeddings

azure_embd_model = AzureOpenAIEmbeddings(
    deployment="text-embedding-3-small",
    azure_endpoint="https://resource-ms-foundary.cognitiveservices.azure.com/",
    openai_api_version="2023-05-15",
    openai_api_key=api_key
)
````

````python
embedded_query = azure_embd_model.embed_query("Hi there")

print("Dimensionality of embedded vector:", len(embedded_query))
print()
print(embedded_query[:15])
````

````python
# Embed list of docs

docs = [
    "Hi there!",
    "Oh, hello!",
    "What's your name?",
    "My friends call me World",
    "Hello World!"
]

embeddings = azure_embd_model.embed_documents(docs)

print("Number of Embeddings:", len(embeddings))

print("Dimensionality of Embeddings:", len(embeddings[0]))
````

````python
from azure.identity import DefaultAzureCredential
from langchain_azure_ai.embeddings import AzureAIOpenAIApiEmbeddingsModel

OPENAI_BASE_URL="https://resource-ms-foundary.services.ai.azure.com/openai/v1"

azure_embd_model = AzureAIOpenAIApiEmbeddingsModel(
	endpoint=OPENAI_BASE_URL,
	credential=DefaultAzureCredential(),
	model="text-embedding-3-small",
)
````

````python
embedded_query = azure_embd_model.embed_query("Hi there")

print("Dimensionality of embedded vector:", len(embedded_query))
print()
print(embedded_query[:15])
````

````python
# Embed list of docs

docs = [
    "Hi there!",
    "Oh, hello!",
    "What's your name?",
    "My friends call me World",
    "Hello World!"
]

embeddings = azure_embd_model.embed_documents(docs)

print("Number of Embeddings:", len(embeddings))

print("Dimensionality of Embeddings:", len(embeddings[0]))
````
