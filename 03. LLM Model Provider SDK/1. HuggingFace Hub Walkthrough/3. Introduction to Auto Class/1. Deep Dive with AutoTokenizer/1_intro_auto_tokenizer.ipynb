```python
import torch
import transformers

print(f"Torch Version: {torch.__version__}")
print(f"Transformer Version: {transformers.__version__}")
```

```python
from transformers import AutoTokenizer

# define the sentence to tokenize
sentence = "Hi, my name is Kanav!"

# load the pretrained tokenizer
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-cased")

print("Vocabulary Size:", tokenizer.vocab_size)
```

```python
# apply the tokenizer to the sentence and extract the token ids
token_ids = tokenizer(sentence).input_ids

print(token_ids)
```

```python
for token in token_ids:
    print(tokenizer.decode(token))
```

## **Visualizing Tokens**

```python
# This is the utility function that will allow us to visually represent tokens 
# A list of colors in RGB for representing the tokens
colors = [
    '102;194;165', '252;141;98', '141;160;203',
    '231;138;195', '166;216;84', '255;217;47'
]

def show_tokens(sentence: str, tokenizer_name: str):
    """ Show the tokens each separated by a different color """

    # Load the tokenizer and tokenize the input
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_name)
    token_ids = tokenizer(sentence).input_ids

    # Extract vocabulary length
    print(f"Vocab length: {len(tokenizer)}")

    # Print a colored list of tokens
    for idx, token in enumerate(token_ids):
        print(
            f'\x1b[0;30;48;2;{colors[idx % len(colors)]}m' +
            tokenizer.decode(token) +
            '\x1b[0m',
            end=' '
        )
```

```python
text = "Hi, my name is Kanav!"

show_tokens(text, "google-bert/bert-base-cased")
```

```python
text = "Hi, my name is Kanav!"

show_tokens(text, "google-bert/bert-base-uncased")
```

```python
text = "Hi, my name is Kanav!"

show_tokens(text, "microsoft/Phi-3-mini-4k-instruct")
```
