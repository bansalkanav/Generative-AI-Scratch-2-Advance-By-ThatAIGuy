## **Github Tool Kit**

Setup:
1. Create Github App:
    - Goto Developer Settings and Create a New App
    - Enter "GitHub App name"
    - Under "Homepage URL", enter your github account url
    - Click "Create Github App"
2. Make sure your app has following repo permissions:
    - Commit statuses (read only)
    - Contents (read only)
    - Issues (read only)
    - Metadata (read only)
    - Pull requests (read only)
3. Get the following environment variables:
    - GITHUB_APP_ID: A unique integer identifier assigned by GitHub to your App, available in the app's general settings.
    - GITHUB_APP_PRIVATE_KEY: The location of your app’s private key .pem file, or the full text of that file as a string.
    - GITHUB_REPOSITORY: The name of the GitHub repository you want your bot to act upon.

```python
# ! pip install PyGithub
```

```python
GITHUB_APP_ID = open("keys/.github_app_id.txt").read()
GITHUB_APP_PRIVATE_KEY = open("keys/.bansalkanav-app-1.2026-04-20.private-key.pem").read()
GITHUB_REPOSITORY = "bansalkanav/GenAI-AgenticAI-From-Zero-to-Production"
```

```python
from langchain_community.utilities.github import GitHubAPIWrapper

github = GitHubAPIWrapper(
    github_app_id=GITHUB_APP_ID,
    github_app_private_key=GITHUB_APP_PRIVATE_KEY,
    github_repository=GITHUB_REPOSITORY, 
)
```

```python
from langchain_community.agent_toolkits.github.toolkit import GitHubToolkit

toolkit = GitHubToolkit.from_github_api_wrapper(github)
```

```python
tools = toolkit.get_tools()

print(f"Number of Tools: {len(tools)}")
print()

for tool in tools:
    print(tool.name)
```

#### **Get Release Tools**

By default, the toolkit does not include release-related tools. You can include them by setting `include_release_tools=True` when initializing the toolkit. This will help include the following tools as well:

- **Get Latest Release** - fetches the latest release from the repository.
- **Get Releases** - fetches the latest 5 releases from the repository.
- **Get Release** - fetches a specific release from the repository by tag name, e.g. v1.0.0. 

```python
from langchain_community.agent_toolkits.github.toolkit import GitHubToolkit

toolkit = GitHubToolkit.from_github_api_wrapper(github, include_release_tools=True)

tools = toolkit.get_tools()

print(f"Number of Tools: {len(tools)}")
print()

for tool in tools:
    print(f"{tool.name}")
```

```python
from langchain_openai import ChatOpenAI

# Setup API Key
f = open('keys/.openai_api_key.txt')
OPENAI_API_KEY = f.read()

chat_model = ChatOpenAI(
    api_key=OPENAI_API_KEY, 
    model="gpt-4o-mini", 
    temperature=1
)
```

```python
from langchain.agents import create_agent

tools=[]
for tool in toolkit.get_tools():
    tool.name = tool.mode
    tools.append(tool)

agent = create_agent(
    model=chat_model,
    tools=tools,
)
```

```python
# response = agent.invoke({"messages": "can you help me list the issues?"})

# for msg in response["messages"]:
#     msg.pretty_print()

for event in agent.stream(
    {"messages": [("user", "can you help me list the issues?")]},
    stream_mode="values",
):
    event["messages"][-1].pretty_print()
```
