We will implement a **data agent** as a multi-agent workflow. This agent will conduct web research using user query and visualize/synthesis the summary of the results. 

We will build a **hierarchical data agent**
1. Start with a **planner**. It will take user query and break it into sub-goals, ready for agent to take action on
2. Then we will pass to the **executor** to execute each step using the sub-agents to accompalish each step in the plan.
    - Web Researcher (Tavily)
    - Chart Generator (Python) and Summarizer
    - Synthesizer
