## **Steps**

1. Login at portal.azure.com
2. Search for "Microsoft Foundry"
3. Create a resource group
4. Go to Foundry Portal
5. Click on "Discover" to explore models
6. Deploy a model - Copy the Deployment Name
7. Copy the following from the Foundry Home Page:
    - Azure OpenAI endpoint
    - API key

```python
with open("keys/.azure_api_key.txt") as file:
    FOUNDRY_API_KEY = file.read()       # Contains the "API key" from Foundry Home Page


with open("keys/.azure_endpoint.txt") as file:
    AZURE_OPENAI_ENDPOINT = file.read() # Contains the "API key" from Foundry Home Page
```

```python
from openai import OpenAI

client = OpenAI(
    base_url=AZURE_OPENAI_ENDPOINT,
    api_key=FOUNDRY_API_KEY,
)

response = client.chat.completions.create(
    model="gpt-4o-mini",              # Go to "Build" > "Models" > "Deployments" - Use your deployment name here
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Explain the concept of transformer algorithm in one line."}
    ]
)

print(response.choices[0].message.content)
```
