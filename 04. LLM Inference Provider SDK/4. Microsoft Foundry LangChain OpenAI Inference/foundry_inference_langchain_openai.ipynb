## **Embeddings**

````python
from langchain_openai import AzureOpenAIEmbeddings

embd_model = AzureOpenAIEmbeddings(
    api_key=FOUNDRY_API_KEY,
    azure_endpoint="https://inst-foundry-evoke.services.ai.azure.com",
    deployment="text-embedding-3-small",
)
````

````python
input_text = "The meaning of life is 42"
vector = embd_model.embed_query(input_text)
print(f"Length of Vector: {len(vector)}")
print(f"First few dimensions: {vector[:3]}")
````

````python
input_texts = ["The meaning of life is 42", "It's a nice day."]
vectors = embd_model.embed_documents(input_texts)
print(f"Number of Vectors: {len(vectors)}")
print(f"Length of each vector: {len(vectors[0])}")
print(f"First few dimensions: {vectors[0][:3]}")
````
